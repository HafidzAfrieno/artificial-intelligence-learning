# **STEP 1: KELAS VECTOR**

```python
class Vector:
    def __init__(self,data):
        self.data = list(data)
        self.size = len(self.data)

    def __reps__(self):
        return f"Vector({self.__delattr__})"

    def __add__(self,other):
        return Vector([a + b for a,b in zip(self.data,other.data)])

    def __sub__(self,other):
        return Vector([a - b for a,b in zip(self.data,other.data)])

    def __mul__(self,scalar):
         return Vector([x * scalar for x in self.data])

    def dot(self,other):
        return sum( a * b for a,b in zip(self.data,other.data))

    def magnitude(self):
        return sum( x ** 2 for x in self.data) ** 0.5

    def normalize(self):
        mag = self.magnitude()
        return Vector([X / mag for X in self.components])
    
    def consine_similarity(self,other):
        return self.dot(other) / (self.magnitude() * other.magnitude())
    
```

# **STEP 2: KELAS MATRIKS DAN OPERASINYA**

```python
class Matrix:
    def __init__(self,data):
        self.data = [list(row) for row in data]
        self.rows = len(self.data)
        self.cols = len(self.data[0])
        self.shape = (self.rows,self.cols)

    def __repr__(self):
        rows_str = "\n  ".join(str(row) for row in self.data)
        return f"Matrix({self.shape}):\n  {rows_str}"

    def __add__(self,other):
        return Matrix([
            [self.data[i][j] + other.data[i][j] for j in range(self.cols)]
            for i in range(self.rows)
        ])

    def __sub__(self,other):
        return Matrix([
            [self.data[i][j] - other.data[i][j] for j in range(self.cols)]
            for i in range(self.rows)
        ])

    def matmul(self,other):
        return Matrix([
            [
                sum(self.data[i][k] * other.data[k][j] for k in range(self.cols))
                for j in range(other.cols)
            ]
            for i in range(self.rows) 
        ])

    def scalar_multiply(self,scalar):
        return Matrix([
            [self.data[i][j] * scalar for j in range(self.cols)]
            for i in range(self.rows)
        ])

    def wise_multiply(self,other):
        return Matrix([
            [self.data[i][j] * other.data[i][j] for j in range(self.cols)]
            for i in range(self.rows)
        ])

    def transpose(self):
        return Matrix([
            [self.data[j][i] for j in range(self.rows)]
            for i in range(self.cols)
        ])

    def determinant(self):
        if self.shape == (1,1):
            return self.data[0][0]
        elif self.shape == (2,2):
            return (self.data[0][0] * self.data[1][1]) - (self.data[0][1] * self.data[1][0])
        det = 0 
        for j in range(self.cols):
            minor = Matrix([
                [self.data[i][k] for k in range(self.cols) if k != j]
                for i in range(1,self.rows)
            ])
            det += ((-1) ** j) * self.data[0][j] * minor.determinant()
        return det

    def inverse_2x2(self):
        det = self.determinant()
        if det == 0:
            raise ValueError("Matrix is singular, no inverse exists")
        return Matrix([
            [self.data[1][1] / det,-self.data[0][1] / det],
            [-self.data[1][0] / det,self.data[0][0] / det]
        ])

    @staticmethod
    def indextity(n):
        return Matrix([
            [1 if i == j else 0 for j in range(n)]
            for i in range(n)
        ])
```

**01.** Verifikasi inversnya. Kalikan A @ A.inverse_2x2() dan pastikan Anda mendapatkan matriks identitas. Cobalah dengan tiga matriks 2x2 yang berbeda. Apa yang terjadi ketika determinannya nol?

```python
class VerifyInverse:
    def calculate_matrix(self, other: list[Matrix]):
        results = []

        for matrix in other:
            determinant = matrix.determinant()
            if abs(determinant) < 1e-10:
                results.append("Matriks tidak memiliki invers")
                continue

            inverse = matrix.inverse_2x2()
            identity = matrix.matmul(inverse)
            results.append(identity)

        return results

matrix_1 = Matrix([[1, 2],[3, 4]])
matrix_2 = Matrix([[2, 1],[5, 3]])
matrix_3 = Matrix([[4, 2],[2, 1]])

verifier = VerifyInverse()
hasil = verifier.calculate_matrix([matrix_1,matrix_2,matrix_3])

for i, result in enumerate(hasil,1):
    print(f"Matriks {i}:")
    print(result)
```

**02.** Implementasikan invers matriks 3x3. Perluas kelas Matrix untuk menghitung invers matriks 3x3 menggunakan metode adjugate. Uji coba terhadap NumPy np.linalg.inv.

```python
import numpy as np
class InversAdjugate:
    def __init__(self,matrix):
        self.matrix = np.array(matrix, dtype=float)
        self.determinan = 0
        self.adjudgate = np.array([])
        self.confactore = np.array([])

    def ajudgate_matrix(self):
        self.determinan = np.linalg.det(self.matrix)
        if np.isclose(self.determinan,0):
            return 'matrik tidak memiliki invers!'
        n = self.matrix.shape[0]
        if self.matrix.ndim != 2 or self.matrix.shape[0] != self.matrix.shape[1]:
            return 'Matriks harus berbentuk persegi!'
        self.confactore = np.zeros((n,n))
        self.confactore = np.array([
            [(-1) ** (i + j) * np.linalg.det(
                np.delete(
                    np.delete(self.matrix, i, axis=0),j, axis=1)
                )
                for j in range(n)
            ]
            for i in range(n)
        ])

        self.adjudgate = self.confactore.T
        invers = self.adjudgate / self.determinan
        return invers

matrix_np = np.array([
    [1,2,3],
    [2,1,5],
    [1,1,2]
])

inv_ajd = InversAdjugate(matrix_np)
result = inv_ajd.ajudgate_matrix()
print(f"invers matrix dari:\n {matrix_np}\n adalah\n{result}")
```

**03.** Bangun jaringan dua lapis. Hanya menggunakan kelas Matrix Anda (tanpa NumPy), buat jaringan saraf dua lapis: input (3) -> tersembunyi (4) -> output (2). Inisialisasi bobot acak, jalankan forward pass, dan verifikasi semua bentuk sudah benar.

```python
class NeuralNetwork(Matrix):
    def __init__(self, input_size, hidden_size, output_size):
        self.W1 = Matrix([
            [random.uniform(-1, 1) for _ in range(input_size)]
            for _ in range(hidden_size)
        ])
        self.W2 = Matrix([
            [random.uniform(-1, 1) for _ in range(hidden_size)]
            for _ in range(output_size)
        ])
        self.b1 = Matrix([[0.1] for _ in range(hidden_size)])
        self.b2 = Matrix([[0.1] for _ in range(output_size)])

    def relu_matrix(self,other):
        return Matrix([[max(0,val) for val in row] for row in other.data])

    def linear_regression(self, weights, input, bias):
        return weights.matmul(input) + bias

    def forward(self,input):
        weights = [self.W1, self.W2]
        biases = [self.b1, self.b2]
        output = input

        for weight, bias in zip(weights, biases):
            Z = self.linear_regression(weight, output, bias)
            output = self.relu_matrix(Z)

        return output

inputs = Matrix([[0.5],[0.8],[0.2]])
nn = NeuralNetwork(input_size=3,hidden_size=4,output_size=2)
final_output = nn.forward(inputs)

print(f'input data:\n {inputs.data}')
print(f'final output:\n {final_output.data}')
```
