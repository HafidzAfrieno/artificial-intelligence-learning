# **Step 1: Vectors From Scratch**

```python
class Vector:
    def __init__(self,components):
        self.components = list(components)
        self.dimension = len(self.components)

    def __add__(self,other):
        return Vector([a + b for a, b in zip(self.components,other.components)])

    def __sub__(self,other):
        return Vector([a - b for a, b in zip(self.components,other.components)])

    def __repr__(self):
        return f"Vector ({self.components})"

    def dot(self,other):
        return sum(a * b for a, b in zip(self.components,other.components))

    def magnitude(self):
        return sum(X**2 for X in self.components) ** 0.5

    def normalize(self):
        mag = self.magnitude()
        return Vector([X / mag for X in self.components])

    def consine_similarity(self,other):
        return self.dot(other) / (self.magnitude() * other.magnitude())
```

```python
a = Vector([1,2,3])
b = Vector([7,8,9])

addition_vector = a + b
subtraction_vector = a - b
dotProduct_vector = a.dot(b)

vector_length_a = a.magnitude()
vector_length_b = b.magnitude()

unit_vector_a = a.normalize()
unit_vector_b = b.normalize()

cosine_similarity_vector = a.consine_similarity(b)
```

```python
print(f"Vektor A                  : {a}")
print(f"Vektor B                  : {b}")
print("-" * 45)
print(f"Penjumlahan (a + b)       : {addition_vector}")
print(f"Pengurangan (a - b)       : {subtraction_vector}")
print(f"Dot Product (a . b)       : {dotProduct_vector}")
print("-" * 45)
print(f"Panjang Vektor A (|a|)    : {vector_length_a:.4f}")
print(f"Panjang Vektor B (|b|)    : {vector_length_b:.4f}")
print("-" * 45)
print(f"Vektor Satuan A (u_a)     : {unit_vector_a}")
print(f"Vektor Satuan B (u_b)     : {unit_vector_b}")
print(f"Cosine Similarity (a, b)  : {cosine_similarity_vector:.4f}")
```

# **Step 2: Matrices from scratch**

```python
class Matrix:
    def __init__(self,rows):
        self.rows = [list(row) for row in rows]
        self.shape = (len(self.rows),len(self.rows[0]))

    def __repr__(self):
        return f'Matrix({self.rows})'

    def __matmul__(self,other):
        if isinstance(other,Vector):
            return Vector([sum(self.rows[i][j] * other.components[j] for j in range(self.shape[1]))
                            for i  in range(self.shape[0])])
        rows = []
        for i in range(self.shape[0]):
            row = []
            for j in range(other.shape[1]):
                row.append(sum(self.rows[i][k] * other.rows[k][j] for k in range(self.shape[1])))
            rows.append(row)
        return Matrix(rows)

    def transpose(self):
        return Matrix([
            [self.rows[j][i] for j in range(self.shape[0])]
            for i in range(self.shape[1])
        ])
        
```

```python
rotation_90 = Matrix([[0,-1],[1,0]])
point = Vector([3,1])

rotated = rotation_90 @ point
print(f"Original: {point}")
print(f"Rotated 90°: {rotated}")
```

**01.** Implementasikan fungsi Vector.angle_between(other) yang mengembalikan sudut dalam derajat antara dua vektor.

```python
import math
class VectorAngle(Vector):
    def __init__(self, components):
        super().__init__(components)
    
    def angle_between(self,other):
        cos_theta = self.consine_similarity(other)
        cos_theta = max(-1.0, min(1.0, cos_theta))
        angle_radians = math.acos(cos_theta)
        return math.degrees(angle_radians)

vec_a = VectorAngle([1,2,3])
vec_b = VectorAngle([4,5,6])

sudut = vec_a.angle_between(vec_b)
print(f"Sudut antara vektor: {sudut:.2f}°")
```

**02.** Buat matriks penskalaan 2D yang menggandakan koordinat x dan melipat gandakan koordinat y, lalu terapkan matriks tersebut ke vektor [1, 1]

```python
class MatrixScaler(Matrix):
    def __init__(self, rows):
        super().__init__(rows)

    def apply_vector(self,other):
        vector_trans = other.transpose()
        return self @ vector_trans
    
matrix_sm = MatrixScaler([[2,0],[0,3]])
matrix_one = MatrixScaler([[1,1]])

hasil = matrix_sm.apply_vector(matrix_one)
print(f'hasil matrixnya :{hasil.rows}')
```

**3.** Diberikan 5 vektor acak berbentuk kata (dimensi 50), temukan dua vektor yang paling mirip menggunakan kesamaan kosinus.

```python
import random
random.seed(42)

class VectorEquation(Vector):
    def __init__(self, components):
        if components and isinstance(components[0], list):
            components = [VectorEquation(c) for c in components]
        super().__init__(components)

    def equation_vector(self):
        count_sim = [
            [i, j, self.components[i].consine_similarity(self.components[j])]
            for i in range(len(self.components))
            for j in range(i + 1, len(self.components))
        ]
    
        return max(count_sim, key=lambda item: item[2])

rand_vector = VectorEquation([[random.gauss(0, 0.2) for _ in range(50)] for _ in range(5)])
hasil = rand_vector.equation_vector()

print(f"Pasangan Vektor dengan sudut terbesar:")
print(f"-> Vektor {hasil[0]} dan Vektor {hasil[1]}")
print(f"-> Sudut (Radian) : {hasil[2]:.4f} rad")
print(f"-> Sudut (Derajat): {math.degrees(hasil[2]):.2f}°")
```

**04.** Verifikasi bahwa keluaran Gram-Schmidt benar-benar ortonormal: periksa bahwa setiap pasangan memiliki hasil perkalian titik 0 dan setiap vektor memiliki magnitudo 1.

```python
class OrthonormalChecker(Vector):
    def __init__(self, components):
        super().__init__(components)
        self.orthonormal = []

    def project(self,vector,other):
        scalar = vector.dot(other) / other.dot(other)
        return Vector([scalar * x for x in other.components])

    def gram_schmidt(self,other:list[Vector]):
        for v in other:
            w = v
            for u in self.orthonormal:
                proj = self.project(w,u)
                w = w - proj
            if w.magnitude() < 1e-9:
                continue
            self.orthonormal.append(w.normalize())
        return self.orthonormal

vector_1 = Vector([1,0,0])
vector_2 = Vector([1,2,1])
vector_3 = Vector([1,1,1])

checker = OrthonormalChecker([])
basis = checker.gram_schmidt([vector_1,vector_2,vector_3])

for i,u in enumerate(basis):
    print(f" u{i+1} = {u}")
    print(f"|u{i+1}| = {u.magnitude():.6f}")

print(f"u1 · u2 = {basis[0].dot(basis[1]):.6f}")
print(f"u1 · u3 = {basis[0].dot(basis[2]):.6f}")
print(f"u2 · u3 = {basis[1].dot(basis[2]):.6f}")
```

**05.** Buat matriks 3x3 dengan rank 2. Verifikasi menggunakan rank()metode tersebut. Kemudian jelaskan objek geometris apa yang direntang oleh kolom-kolom tersebut.

```python
class MatrixRankVerifier(Matrix):
    def __init__(self, components):
        super().__init__(components)
        self.n = len(self.rows)
        self.dim = len(self.rows[0])
        self.mat = Matrix([row[:] for row in self.rows])

    def rank(self):
        rows = [row[:] for row in self.mat.rows]
        rank = 0
        for col in range(self.dim):
            pivot = None
            for row in range(rank, self.n):
                if abs(rows[row][col]) > 1e-10:
                    pivot = row
                    break
            if pivot is None:
                continue
            rows[rank], rows[pivot] = rows[pivot], rows[rank]
            scale = rows[rank][col]
            rows[rank] = [x / scale for x in rows[rank]]
            for row in range(self.n):
                if row != rank and abs(rows[row][col]) > 1e-10:
                    factor = rows[row][col]
                    rows[row] = [
                        rows[row][j] - factor * rows[rank][j]
                        for j in range(self.dim)
                    ]
            rank += 1
        return rank

matrix = MatrixRankVerifier([
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 1]
])
hasil = matrix.rank()

print("Rank Matrix:", hasil)
if hasil == 2:
    print("Verifikasi berhasil: Rank matriks adalah 2.")
else:
    print("Verifikasi gagal.")
```
