# **STEP 1 : TURUNAN CALCULUS**

```python
class Calculus:
    def __init__(self,h=1e-7):
        self.h = h

    def derivative(self,f,x):
        return (f(x + self.h) - f(x - self.h)) / (2 * self.h)

    def partial_gradient(self,f,point):
        gradient = []
        for i in range(len(point)):
            point_plus = list(point)
            point_minus = list(point)
            point_plus[i] += self.h
            point_minus[i] -= self.h
            partial = (f(point_plus) - f(point_minus)) / (2 * self.h)
            gradient.append(partial)
        return gradient

    def gradient_hession_2d(self,f,x,y):
        fxx = (f( x + self.h, y ) - 2 * f(x,y) + f(x - self.h,y)) / (self.h ** 2) 
        fyy = (f( x , y + self.h) - 2 * f(x,y) + f(x,y - self.h)) / (self.h ** 2)
        fxy = (f(x + self.h, y + self.h) - f(x + self.h, y - self.h) - f(x - self.h, y + self.h) + f(x - self.h, y - self.h)) / (4 * self.h ** 2)
        return [[fxx,fxy],[fxy,fyy]]
```

```python
calc = Calculus()

def f1(x):
    return x ** 2

number = [-2,-1,0,1,2,3]

for x in number:
    numerical = calc.derivative(f1,x)
    analytical = 2 * x
    print(f"x= {x:2d}  f'(x) numerical={numerical:.2f}  analytical={analytical:.2f}")
```

# **STEP 2 : TURUNAN PARISAL DAN GRADIENT**

```python
def f2(point):
    x,y = point
    return x**2 + 3*x*y + y**2

points = [[1,2],[2,3],[3,4]]

for a in points:
    x,y = a
    numerical = calc.partial_gradient(f2,a)
    analytical = [2*x + 3*y, 3*x + 2*y]
    print(
        f"point= ({x:2d}, {y:2d})  "
        f"numerical=[{numerical[0]:.2f}, {numerical[1]:.2f}]  "
        f"analytical=[{analytical[0]:.2f}, {analytical[1]:.2f}]"
    )
```

# **STEP 3 : APLIKASI GRADIENT UNTUK UPDATE NILAI BOBOT**

```python
weight = 5
learning_rate = 0.1

def func_grad(x):
    return x ** 2

for step in range(5):
    grad = calc.derivative(func_grad, weight)
    weight = weight - learning_rate * grad
    print(f"step {step:2d}  weight={weight:8.4f}  new weight={weight**2:10.6f}")
```

# **STEP 4 : PENURUNAN GRADIENT PADA FUNGSI 2D**

```python
def f_2d(point):
    x, y = point
    return x**2 + y**2

point = [4.0, 3.0]
lr = 0.1
for step in range(30):
    grad = calc.partial_gradient(f_2d, point)
    point = [p - lr * g for p, g in zip(point, grad)]
    loss = f_2d(point)
    if step % 5 == 0 or step == 29:
        print(f"step {step:2d}  point=({point[0]:7.4f}, {point[1]:7.4f})  f={loss:.6f}")
```

# **STEP 7 : DERET TAYLOR**

```python
import math

def taylor_approx(f, f_prime, f_double_prime, x0, h, order=2):
    result = f(x0)
    if order >= 1:
        result += f_prime(x0) * h
    if order >= 2:
        result += 0.5 * f_double_prime(x0) * h ** 2
    return result

x0 = 0.0
for h in [0.1, 0.5, 1.0, 2.0]:
    true_val = math.sin(h)
    t1 = taylor_approx(math.sin, math.cos, lambda x: -math.sin(x), x0, h, order=1)
    t2 = taylor_approx(math.sin, math.cos, lambda x: -math.sin(x), x0, h, order=2)
    print(f"h={h:.1f}  sin(h)={true_val:.4f}  order1={t1:.4f}  order2={t2:.4f}")
```

# **STEP 8 : APLIKASI GRADIENT PADA JARIANGAN SARAF**

```python

import numpy as np

class NeuralNetwork(Calculus):
    def __init__(self, lr=0.01, epoch=2):
        super().__init__()
        self.learning_rate = lr
        self.epoch = epoch
        self.weight = np.random.randn()
        self.bias = np.random.randn()

    def regression(self, X, weight=None, bias=None):
        weight = self.weight if weight is None else weight
        bias = self.bias if bias is None else bias

        return weight * X + bias

    @staticmethod
    def count_error(pred, y):
        return pred - y

    def forward(self, X, y):
        X = np.asarray(X, dtype=float)
        y = np.asarray(y, dtype=float)

        pred = self.regression(X)
        error = self.count_error(pred, y)

        loss = np.mean(error ** 2)
        dw = self.derivative( lambda w: np.mean((self.regression(X, weight=w) - y) ** 2), self.weight)
        db = self.derivative(lambda b: np.mean((self.regression(X, bias=b)  - y) ** 2),self.bias)

        self.weight -= self.learning_rate * dw
        self.bias -= self.learning_rate * db

        return loss

    def train(self, X, y):
        for i in range(self.epoch):
            loss = self.forward(X, y)

            print(
                f"Epoch {i+1}: "
                f"loss={loss:.6f}, "
                f"w={self.weight:.4f}, "
                f"b={self.bias:.4f}"
            )
```

```python
nn = NeuralNetwork(epoch=10)
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

nn.train(X=x,y=y)
```
